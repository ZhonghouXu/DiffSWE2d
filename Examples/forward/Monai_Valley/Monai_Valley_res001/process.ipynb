```python
gauge0 = np.loadtxt("gauge_0.txt",skiprows=2)
gauge1 = np.loadtxt("gauge_1.txt",skiprows=2)
gauge2 = np.loadtxt("gauge_2.txt",skiprows=2)

mea = np.loadtxt("MonaiValley_WaveGages.txt",skiprows=1)
```

```python
fig, ax = plt.subplots(3, 1, figsize=(8, 6), sharex=True)
fig.subplots_adjust(hspace=0)

ax[0].plot(mea[:, 0], mea[:, 1] / 100, color='red', linestyle='--', label='Measured')
ax[0].plot(gauge0[:, 0], gauge0[:, 3], color='k', linestyle='-',label='Model')
ax[0].text(0.02, 0.95, 'gauge0', transform=ax[0].transAxes, fontsize=10, va='top')

ax[1].plot(mea[:, 0], mea[:, 2] / 100, color='red', linestyle='--', label='Measured')
ax[1].plot(gauge1[:, 0], gauge1[:, 3], color='k', linestyle='-',label='Model')
ax[1].text(0.02, 0.95, 'gauge1', transform=ax[1].transAxes, fontsize=10, va='top')

ax[2].plot(mea[:, 0], mea[:, 3] / 100, color='red', linestyle='--', label='Measured')
ax[2].plot(gauge2[:, 0], gauge2[:, 3], color='k', linestyle='-',label='Model')
ax[2].text(0.02, 0.95, 'gauge2', transform=ax[2].transAxes, fontsize=10, va='top')

for a in ax:
    a.set_xlim(0, 30)
    a.set_ylim(-0.02, 0.05)
    a.set_xlabel('Time (s)', fontsize=12)
    a.set_ylabel('Water Level (m)', fontsize=12)

ax[0].legend(loc='upper right', fontsize=10)
ax[1].legend(loc='upper right', fontsize=10)
ax[2].legend(loc='upper right', fontsize=10)

fig.savefig("wave_gauges_comparison.png", dpi=300, bbox_inches='tight')
```
