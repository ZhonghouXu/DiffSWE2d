```python
gauge0 = np.loadtxt("gauge_0.txt",skiprows=2)
gauge1 = np.loadtxt("gauge_1.txt",skiprows=2)
gauge2 = np.loadtxt("gauge_2.txt",skiprows=2)
gauge3 = np.loadtxt("gauge_3.txt",skiprows=2)
gauge4 = np.loadtxt("gauge_4.txt",skiprows=2)
gauge5 = np.loadtxt("gauge_5.txt",skiprows=2)
```

```python
simg = np.column_stack([
    gauge0[:, 0], 
    gauge0[:, 3], 
    gauge1[:, 3], 
    gauge2[:, 3], 
    gauge3[:, 3], 
    gauge4[:, 3], 
    gauge5[:, 3]
])
```

```python
simg
```

```python
# 1. Create the new time array from 0 to 30 seconds with 0.2s steps
t_orig = simg[:, 0]
t_new = np.arange(0, 30, 0.2) 

# 2. Start a list to hold your new columns, beginning with the new time column
interpolated_cols = [t_new]

# 3. Interpolate each gauge column (indices 1 through 6)
for i in range(1, simg.shape[1]):
    gauge_interp = np.interp(t_new, t_orig, simg[:, i])
    interpolated_cols.append(gauge_interp)

# 4. Stack them back into a single 2D array
simg_interp = np.column_stack(interpolated_cols)
```

```python
# Save the entire 2D array to a single text file
np.savetxt(
    "sim_gauges.txt", 
    simg_interp, 
    fmt="%.6e",          # Scientific notation with 6 decimal places (matches typical gauge output)
    delimiter=" ",       # Space separated
    header="Time Gauge0 Gauge1 Gauge2 Gauge3 Gauge4 Gauge5", 
    comments=""          # Prevents a '#' from being added to the start of the header
)
```

```python
plt.plot(tide_data[:,0],tide_data[:,1],'k',label='Input wave')
plt.plot(simg[:,0],simg[:,1],'r',label='Gauge 0')
plt.plot(simg[:,0],simg[:,2],'g',label='Gauge 1')
plt.plot(simg[:,0],simg[:,3],'b',label='Gauge 2')
plt.plot(simg[:,0],simg[:,4],'c',label='Gauge 3')
plt.plot(simg[:,0],simg[:,5],'m',label='Gauge 4')
plt.plot(simg[:,0],simg[:,6],'y',label='Gauge 5')  

plt.xlim(0, 30)
plt.xlabel('Time (s)', fontsize=12)
plt.ylabel('Water Level (m)', fontsize=12)
plt.legend(fontsize=10)
#plt.show()
plt.savefig("wave_gages_data.png", dpi=300, bbox_inches='tight')
```
